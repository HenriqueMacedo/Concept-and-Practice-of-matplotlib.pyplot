```python
# ETAPA 2 - FÓRMULA COMPOSTA - UM CASO DE SUBDIVISÕES
# Determine a solução usando a fórmula composta, considerando o número de 6 subdivisões do trapézio (total de 7 pontos).
import numpy as np

print("\n==========================")
print("O que eu fiz em sala de aula")
a = 3
b = 3.6
n = 6 # Número de subdivisões
h = (b - a) / n

# Algoritmo de Integral
x = a
funcao = 1 / x # algoritmo f(x)
area_total = 0
i = 0

while i < n:
    x = x + h
    y = 1 / x # algoritmo f(x)
    area_trapezio = (funcao + y) * (h / 2)
    area_total = area_total + area_trapezio
    funcao = y
    i = i + 1

print(f"O resultado da integral é:{area_total: .4f}")
```

```python
# ETAPA 3 - ERRO DO MÉTODO DE INTEGRAÇÃO
# Calcular o erro de truncamento do método de integração
from sympy import diff, Symbol

x_int = np.array([a, b])
max_valor = None
xs = Symbol('x')
fun = (1/xs)**2 

for i in x_int:
    derivada2 = diff(fun, xs, 2)
    num = abs(derivada2.subs(xs, i))
    if max_valor is None or num > max_valor:
        max_valor = num

print(f"O valor máximo é:{max_valor: .4f}")
erro = (6*((h)**3) / 12) * max_valor
print(f"O erro de truncamento é:{erro: .9f}")

# ===============================================================================
# Correção que foi mandada por e-mail
import numpy as np
from sympy import diff, Symbol
import matplotlib.pyplot as plt

print("\n==========================")
print("Valor da correção do que foi enviado por e-mail")
def f(x):
    return 1/x

a = 3.0
b = 3.6
n = 6
h = (b - a) / n
x = a
funcao = f(x)
area_total = 0
i = 0

while i < n:
    x_proximo = x + h
    y = f(x_proximo)
    area_trapezio = ((funcao + y) / 2) * h
    area_total = area_total + area_trapezio
    funcao = y
    x = x_proximo
    i = i + 1

print(f"O valor da área é:{area_total: .4f}")

# ERRO
xint = np.array([a, b])
max_valor = None
for i in xint:
    xs = Symbol('x')
    fun = 1/xs
    derivada = diff(fun, xs, 2)
    num = 2 / (i**3)
    if max_valor is None or num > max_valor:
        max_valor = num

Erro = (6*((h)**3) / 12) * max_valor
print(f"O valor do erro é:{Erro: .9f}")
```
