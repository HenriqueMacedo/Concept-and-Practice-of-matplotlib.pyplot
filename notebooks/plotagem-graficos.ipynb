```python
import matplotlib.pyplot as plt
import numpy as np

# Primeiro conjunto de gráficos: sen(x) e sen(4x)
x = np.linspace(-40, 40, 500)

plt.figure(figsize=(8, 6))

plt.subplot(2, 1, 1)
plt.plot(x, np.sin(x), "--r")
plt.title("Gráficos de sen(x) e sen(4x)")
plt.ylabel("sen(x)")

plt.subplot(2, 1, 2)
plt.plot(x, np.sin(4*x), ":b")
plt.xlabel("x")
plt.ylabel("sen(4x)")

plt.tight_layout()
plt.show()
```

```python
# Segundo gráfico: Função Exponencial Amortecida
y = np.exp(-0.25*x) * np.sin(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, ":m", label="Gráfico de f(x)")
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"$f(x) = e^{-0.25x} \sin(x)$")
plt.xlim(1, 10)
plt.ylim(-1, 2)
plt.grid(alpha=0.1)
plt.show()
```

```python
# Terceiro conjunto: Subplots comparativos
x = np.linspace(-20, 20, 100)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.title("Curva 1", fontsize=14)
plt.plot(x, np.sin(x)/x)
plt.xlabel("x1")
plt.ylabel("sen(x)/x")

plt.subplot(1, 2, 2)
plt.title("Curva 2", fontsize=14)
plt.plot(x, x*np.sin(x))
plt.xlabel("x2")
plt.ylabel("x * sen(x)")

plt.tight_layout()
plt.show()
```
